# ***Data Cleaning***

```python
#DROP DUPLICATES
df.drop_duplicates(inplace = True)
#CONVERTING NEGATIVE OR UNREALISTIC VALUE INTO REALISTIC VALUE
df['Physical_Activity_Hours'] = df['Physical_Activity_Hours'].clip(lower = 0)
df['Mental_Health_Score'] = df['Mental_Health_Score'].clip(lower = 0)
df['Sleep_Hours_Per_Night'] = df['Sleep_Hours_Per_Night'].clip(lower=0)
```

# ***`FEATURE ENGENEERING`***

```python
top_country = df['Country'].value_counts().index[0:10].to_list()
```

```python
def group_countries(country):
  if country in top_country:
    return country
  else:
    return 'Other'
```

```python
df['Grouped_country'] = df['Country'].apply(group_countries)
display(df['Grouped_country'].value_counts())
```

# ***`TRAIN TEST SPLIT`***

```python
from sklearn.model_selection import train_test_split


skwes_col = ['Study_Hours']
other_col = ['Age','Avg_Daily_Usage_Hours','Daily_Unlocks','Physical_Activity_Hours','Sleep_Hours_Per_Night']
ordinal_col = ['Stress_Level']
normal_col = ['Most_Used_Platform','Gender','Academic_Level','Purpose_Of_Use','Grouped_country']

# Combine into a single flat list
feature_col = skwes_col + other_col + ordinal_col + normal_col

x = df[feature_col]
y = df['Mental_Health_Score']



x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.30,random_state = 42)
```

# ***`PREPROCESSING USING COLUMN TRANSFORMER`***

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder,OrdinalEncoder,FunctionTransformer
#for skewed
skew_pipeline = Pipeline(steps=[
    ('log_transform',FunctionTransformer(np.log1p)),
    ('scale',StandardScaler())
])

#for numeric
plain_numeric_pipeline = Pipeline(steps=[
    ('scale',StandardScaler())
])

#ordinal_pipeline
ordinal_pipeline = Pipeline(steps=[
    ('encode',OrdinalEncoder(categories=[['Low','Medium','High','Very High']])),
])

#nominal_pipeline
nominal_pipeline = Pipeline(steps=[
    ('encode',OneHotEncoder(handle_unknown="ignore"))
])

column_transformer = ColumnTransformer(transformers=[
    ("Skewed_Pipeline",skew_pipeline, skwes_col),
    ("Numeric_Pipeline",plain_numeric_pipeline, other_col),
    ("Ordinal_Pipeline",ordinal_pipeline, ordinal_col),
    ("Nominal_Pipeline",nominal_pipeline, normal_col)
])
```

# ***PIPELINE VISUALISATION***

```python
from sklearn import set_config
from IPython.display import display

set_config(display="diagram")

display(column_transformer)
```

# ***`BUILDING PIPELINE`***

# ***`MODEL BUILDING`***

# ***`Linear Regression`***

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
```

```python
lr_pipeline = Pipeline(steps=[
    ("preprocessor",column_transformer),
    ("regressor",LinearRegression())
])
lr_pipeline.fit(x_train,y_train)
lr_pred = lr_pipeline.predict(x_test)
lr_pred_train = lr_pipeline.predict(x_train)



lr_r2_test = r2_score(y_test,lr_pred)
lr_r2_train = r2_score(y_train,lr_pred_train)
lr_mse_test = mean_squared_error(y_test,lr_pred)
```

```python
print(f"Accuracy of train data is {lr_r2_train}")
print(f"Accuracy of test data is {lr_r2_test}")
print(f"MSE of test data is {lr_mse_test}")
```

# ***`Random Forest`***

```python
from sklearn.ensemble import RandomForestRegressor
rf_pipeline = Pipeline(steps=[
    ("preprocessor",column_transformer),
    ("Random_Forest",RandomForestRegressor(random_state=42,n_estimators=100))
])

rf_pipeline.fit(x_train,y_train)
rf_pred = rf_pipeline.predict(x_test)
rf_pred_train = rf_pipeline.predict(x_train)


rf_r2_test = r2_score(y_test,rf_pred)
rf_r2_train = r2_score(y_train,rf_pred_train)
rf_mae_test = mean_absolute_error(y_test,rf_pred)


print(f"Accuracy of train data is {rf_r2_train}")
print(f"Accuracy of test data is {rf_r2_test}")
print(f"MAE of test data is {rf_mae_test}")

```

# ***`HYPERPARAMEATER TUNING THE RANDOM FOREST`***

```python

from sklearn.model_selection import GridSearchCV
from sklearn.metrics import r2_score

param_grid = {
    'Random_Forest__max_depth': [5, 8, 10, 15],
    'Random_Forest__min_samples_split': [5, 10, 20],
    'Random_Forest__min_samples_leaf': [2, 4, 8],
    'Random_Forest__max_features': [0.5, 0.8]
}

grid_search = GridSearchCV(
    estimator=rf_pipeline,
    param_grid=param_grid,
    cv=3,
    scoring='r2',
    n_jobs=1,
    verbose=1
)

grid_search.fit(x_train, y_train)

best_regularized_model = grid_search.best_estimator_

train_pred = best_regularized_model.predict(x_train)
test_pred = best_regularized_model.predict(x_test)

print("Best parameters:", grid_search.best_params_)
print("Train R²:", r2_score(y_train, train_pred))
print("Test R²:", r2_score(y_test, test_pred))
print("Best CV R²:", grid_search.best_score_)


```

```python

from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# Best tuned pipeline
rf_best_pipeline = grid_search.best_estimator_

# Predictions on test data
rf_best_pred = rf_best_pipeline.predict(x_test)

# Evaluation metrics
rf_best_r2 = r2_score(y_test, rf_best_pred)
rf_best_mae = mean_absolute_error(y_test, rf_best_pred)
rf_best_mse = mean_squared_error(y_test, rf_best_pred)
rf_best_rmse = rf_best_mse ** 0.5

print("Best CV R²:", grid_search.best_score_)
print("Tuned Random Forest Test R²:", rf_best_r2)
print("Tuned Random Forest MAE:", rf_best_mae)
print("Tuned Random Forest MSE:", rf_best_mse)
print("Tuned Random Forest RMSE:", rf_best_rmse)
```

# ***`Checking Before And After`***

```python

print("Original Test R²:", rf_r2_test)
print("Tuned Test R²:", rf_best_r2)

improvement = (rf_best_r2 - rf_r2_test) * 100
print(f"R² improvement: {improvement:.2f} percentage points")
```
